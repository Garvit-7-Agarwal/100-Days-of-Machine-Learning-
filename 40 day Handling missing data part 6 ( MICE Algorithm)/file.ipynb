```python
import pandas as pd 
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/1000)
# Read the CSV → select 4 numerical columns → divide all values by 1000 → round the values → store the result in df.
np.random.seed(42)
# you will get the same random selection each time you run the code, assuming the same NumPy/Pandas behaviour. # similar to random_state=42
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.head()
```

```python
# Step1 -> impute all the missing value with the mean of respective columns 
df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
# 0th iteration 
df0.head()
```

```python
# remove the col1 imputed value 
df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python
# using column 1 and 2 and row 0,2,3,4 for training the model 
X = df1.iloc[[0,2,3,4],1:3]
X
```

```python
y = df1.iloc[[0,2,3,4],0]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:3].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 25.788
df1
```

```python
# Remove the col2 imputed value

df1.iloc[3,1] = np.nan

df1
```

```python
X = df1.iloc[[0,1,2,4],[0,2]]
X
```

```python
y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[3,[0,2]].values.reshape(1,2))
```

```python
df1.iloc[3,1] = 68.96
```

```python
df1
```

```python
# Remove the col3 imputed value
df1.iloc[4,-1] = np.nan

df1
```

```python
X = df1.iloc[0:4,0:2]
X
```

```python
y = df1.iloc[0:4,-1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[4,0:2].values.reshape(1,2))
```

```python
df1.iloc[4,-1] = 46.228
```

```python
# After 1st Iteration
df1
```

```python
# Subtract 0th iteration from 1st iteration

df1 - df0
```

```python
df2 = df1.copy()

df2.iloc[1,0] = np.nan

df2
```

```python
X = df2.iloc[[0,2,3,4],1:3]
y = df2.iloc[[0,2,3,4],0]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[1,1:].values.reshape(1,2))
```

```python
df2.iloc[1,0] = 25.644
```

```python

df2.iloc[3,1] = np.nan
X = df2.iloc[[0,1,2,4],[0,2]]
y = df2.iloc[[0,1,2,4],1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[3,[0,2]].values.reshape(1,2))
```

```python

df2.iloc[3,1] = 68.85
```

```python
df2.iloc[4,-1] = np.nan

X = df2.iloc[0:4,0:2]
y = df2.iloc[0:4,-1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[4,0:2].values.reshape(1,2))
```

```python
df2.iloc[4,-1] = 41.85
```
