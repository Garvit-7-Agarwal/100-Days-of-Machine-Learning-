```python
from sklearn.datasets import make_regression 
import numpy as np 
```

```python
X,y = make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
# n_samples=4 → 4 data points/rows
# n_features=1 → X has only 1 column
# n_targets=1 → one y value for each data point
```

```python
import matplotlib.pyplot as plt 
plt.scatter(X,y)
```

```python
X.shape
```

```python
y.shape
```

```python
# Let apply OLS
from sklearn.linear_model import LinearRegression 
```

```python
reg = LinearRegression()
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
# Now we will aplly the gradient descent and we will try to reach near to 26.15 
```

```python
plt.plot(X,reg.predict(X),color='red') # gives best fit line 
plt.scatter(X,y)
```

```python
# let apply graident descent assuming slope is constant m = 78.35
# and let's assume that starting value of b is 0
y_pred = (X*78.35+100).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='green',label='b=100')
plt.legend()
plt.show()
```

```python
# calculating the new_intercept 

b_new = b-step_size
b_new
```
