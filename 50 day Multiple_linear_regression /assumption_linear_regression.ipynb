```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv('data.csv')
```

```python
df.head()
```

```python
X = df.iloc[:,0:3].values
y = df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

model.fit(X_train,y_train)
```

```python
# Residual 

y_pred = model.predict(X_test)
residual = y_test-y_pred
```

## 1). Linear relationship

```python
fig,(ax1,ax2,ax3) = plt.subplots(ncols=3,figsize=(12,2.5))

ax1.scatter(df['feature1'],df['target'])
ax1.set_title("Feature 1")
ax2.scatter(df['feature2'],df['target'])
ax2.set_title("Feature 2")
ax3.scatter(df['feature3'],df['target'])
ax3.set_title("Feature 3")
```

## 2. Multicollinearity

```python
X_train.shape[1],X_train.shape
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor 

vif = []

for i in range(X_train.shape[1]):
    vif.append(variance_inflation_factor(X_train,i)) 

# variance_inflation_factor calculates the multicollinearity for each feature in X_train if the value of variance_inflation_factor is around 1 then it is ok but if it is around 5 then there is a relationship between the features 
```

```python
pd.DataFrame({'vif':vif},index=df.columns[0:3]).T
```

```python
# Another technique 
sns.heatmap(df.iloc[:,0:3].corr(),annot=True)
```
